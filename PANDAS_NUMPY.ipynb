```python
import pandas as pd
inicio = int(input('Introduce el año inicial: '))
final = int(input('Introduce el año final: '))
ventas = {}
for i in range(inicio, final+1):
    if i % 2 == 0:
        ventas[i] = float(input('Introduce las ventas del año ' + str(i) +': '))
serie_ventas = pd.Series(ventas)
serie_ventas_con_descuento = serie_ventas * 0.95
print("\nDatos de ventas originales:")
display(serie_ventas)
print("\nDatos de ventas con 5% de descuento:")
display(serie_ventas_con_descuento)
```

Ejercicio 2: Calculo de estadisticas basicas: media, mediana, desviación estandar- numpy para el dataframe

```python
import pandas as pd
import numpy as np
#crear un dataframe
data = {'Nombre': ['Jose', 'Juan', 'Maria', 'Pedro'], 'Edad': [32, 28, 19, 21], 'Ciudad': ['Huancayo', 'Lima', 'Piura', 'Trujillo']}
df = pd.DataFrame(data)

#Calcular la media utilizando numpy
media_edad = np.mean(df['Edad'])
print(media_edad)

#Calcular la mediana utilizando numpy
mediana_edad = np.median(df['Edad'])
print(mediana_edad)

#Calcular la desviación estandar utilizando numpy
desviacion_estandar_edad = np.std(df['Edad'])
print(desviacion_estandar_edad)

#Calcular la máxima edad utilizando numpy
max_edad = np.max(df['Edad'])
print(max_edad)

#Calcular la mínima edad utilizando numpy
min_edad = np.min(df['Edad'])
print(min_edad)
```
